```python
import os
import sqlite3

# Configurazione del percorso centralizzato del database
DB_DIR = os.path.join("..", "data")
DB_PATH = os.path.join("..", "data", "platformhero_mirror.db")
# Creazione della cartella 'data' se non esiste
os.makedirs(DB_DIR, exist_ok=True)

print(f" Percorso database specchio configurato: {DB_PATH}")
```

```python
# Connessione al database (crea il file .db se non esiste)
conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

# Creazione della tabella per i log normalizzati (STEP 5)
cursor.execute("""
CREATE TABLE IF NOT EXISTS logs_normalized (
    id TEXT PRIMARY KEY,
    external_id TEXT UNIQUE,
    ticket TEXT,
    client TEXT,
    area TEXT,
    category TEXT,
    raw_markdown TEXT,
    structured_json TEXT,
    usage_json TEXT,
    created_at TEXT
);
""")
```

```python

#  Creazione della tabella per le valutazioni del Judge (STEP 7)
cursor.execute("""
CREATE TABLE IF NOT EXISTS evaluations (
    id TEXT PRIMARY KEY,
    log_id TEXT,
    overall_score REAL,
    technical_score REAL,
    completeness_score REAL,
    business_score REAL,
    consistency_score REAL,
    feedback TEXT,
    issues TEXT,
    created_at TEXT,
    FOREIGN KEY(log_id) REFERENCES logs_normalized(id)
);
""")

conn.commit()
conn.close()
print(" Struttura del database inizializzata correttamente!")
print(" -> Tabella 'logs_normalized' creata o già esistente.")
print(" -> Tabella 'evaluations' creata o già esistente.")
```

```python
conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS chat_logs (
    id TEXT PRIMARY KEY,
    assistant_id TEXT,
    chat_id TEXT,
    message_count INTEGER,
    messages_json TEXT,
    created_at TEXT
);
""")

# Questo è il comando fondamentale che scrive i dati sul file!
conn.commit()
conn.close()

print("Tabella 'chat_logs' creata e salvata stabilmente sul disco!")
```
